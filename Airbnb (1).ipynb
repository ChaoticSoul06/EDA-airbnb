```python
df = pd.read_csv(file_path, low_memory=False)
```

```python
df.columns
```

```python
print(df.isnull().sum())
```

```python
#convert last review to datetime to handle errors, coerce replaces values with nan instead of raisning error
df['last review']=pd.to_datetime(df['last review'], errors='coerce')
#fill missing values
df.fillna({'reviews per month':0,'last review':df['last review'].min()}, inplace=True)
#drop records with missing name or hostname
df.dropna(subset=['NAME', 'host name'], inplace=True)
```

```python
print(df.head())
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
#remove dollar signs and convert to float
df['price'] = df.['price'].replace('[\$,]','', regex=True).astype(float)
df['service fee'] = df['service fee'].replace('[\$,]','', regex=True).astype(float)
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50, kde=True)
plt.title('Distribution of Listing Prices')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.show
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='room type', data=df , color='hotpink')
plt.title('Room Type Distribution')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.countplot(y='neighbourhood group', data=df, color="lightgreen", order=df['neighbourhood group'].
value_counts().index)
plt.title('Number of listings by neighborhood group')
plt.xlabel('Count')
plt.ylabel('Neighborhood Group')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='room type', y='price', hue='room type', data=df, palette='Set2')
plt.title('Price vs. Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price ($)')
plt.legend(title='Room Type')
plt.show()
```

```python
df['last review'] = pd.to_datetime(df['last review'])
reviews_over_time = df.groupby(df['last review'].dt.to_period('M')).size()

plt.figure(figsize=(12, 6))
reviews_over_time.plot(kind='line',color='red')
plt.title('Number of Reviews Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')
plt.show()
```
